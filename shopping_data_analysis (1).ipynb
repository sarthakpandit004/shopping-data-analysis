## 1. loading the data

```python
df = pd.read_csv('shopping_dataset.csv')
print('done')
```

## 3. missing values

saw some nulls in the info() output, lets see exactly where

```python
df.isnull().sum()
```

```python
# age has 5 missing - filling with median makes more sense than mean here
# (mean can get skewed by outliers)
df['age'].fillna(df['age'].median(), inplace=True)

# product_category and product_name - just putting unknown for now
df['product_category'].fillna('Unknown', inplace=True)
df['product_name'].fillna('Unknown', inplace=True)
```

```python
# checking if anything is left
df.isnull().sum()
```

good, no more nulls

## 4. filtering and selecting columns

```python
# just selecting the columns that actually matter for analysis
cols = ['order_id', 'customer_name', 'product_category', 'product_name', 'quantity', 'price', 'payment_method', 'city']
df[cols].head()
```

```python
# filtering - only electronics
df[df['product_category'] == 'Electronics']
```

```python
# orders where price is more than 1500
df[df['price'] > 1500][['order_id', 'customer_name', 'product_name', 'price']]
```

```python
# female customers who paid with upi
df[(df['gender'] == 'Female') & (df['payment_method'] == 'UPI')]
```

## 5. removing duplicates

```python
print('before:', df.shape)
print('duplicate rows:', df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)

print('after:', df.shape)
```

## 6. adding total_amount column

price x quantity to get what each order actually cost

```python
df['total_amount'] = df['price'] * df['quantity']

df[['product_name', 'price', 'quantity', 'total_amount']].head(10)
```

```python
# quick look at revenue by category
df.groupby('product_category')['total_amount'].sum().sort_values(ascending=False)
```

```python
# some basic stats
print('total revenue:', df['total_amount'].sum())
print('avg order value:', round(df['total_amount'].mean(), 2))
print('highest order:', df['total_amount'].max())
```

## 7. saving cleaned file

```python
df.to_csv('shopping_dataset_cleaned.csv', index=False)
print('saved')
```
